```
# 3. Is the game video in Drive, complete and readable?
import os, cv2
p = '/content/drive/MyDrive/shift-tracker/clips/bu_game.mp4'
EXPECTED = 2645451569  # size in bytes of the file on the Mac
size = os.path.getsize(p)
cap = cv2.VideoCapture(p)
frames = cap.get(cv2.CAP_PROP_FRAME_COUNT); fps = cap.get(cv2.CAP_PROP_FPS)
print('size matches:', size == EXPECTED, size)
print(f'{cap.get(3):.0f}x{cap.get(4):.0f}, duration {frames/fps/60:.1f} min (expect about 126.8)')
ok, _ = (cap.set(cv2.CAP_PROP_POS_FRAMES, frames-200), cap.read())[1]
print('last part readable:', ok)
print('SAFE TO DELETE FROM MAC' if size == EXPECTED and ok else 'DO NOT DELETE - something is wrong')
```
